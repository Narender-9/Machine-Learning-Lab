```python
# @title Exploratory Data Analysis and Statistical Insights - Titanic Dataset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic Dataset
df = sns.load_dataset("titanic")

# Display First 5 Rows
print(df.head())

# Dataset Information
print(df.info())

# Descriptive Statistics
print(df.describe())

# Check Missing Values
print(df.isnull().sum())
```

```python
# Display the correlation matrix for numerical columns
print(df.corr(numeric_only=True))
```

```python
# Visualize the number of passengers who survived and did not survive
sns.countplot(x='survived', data=df)
plt.title("24EU02012[AI&ML]-Survival Count")
plt.show()
```

```python
# Visualize the distribution of passengers by gender
sns.countplot(x='sex', data=df)
plt.title("24EU02012[AI&ML]-Gender Distribution")
plt.show()
```

```python
# Visualize the distribution of passenger classes
sns.countplot(x='pclass', data=df)
plt.title("24EU02012[AI&ML]-Passenger Class Distribution")
plt.show()
```

```python
# Display the distribution of passenger ages
df['age'].hist(bins=20)
plt.title("24EU02012[AI&ML]-Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Display the distribution of passenger fares
df['fare'].hist(bins=20)
plt.title("24EU02012[AI&ML]-Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()
```

```python
# Display the correlation heatmap for numerical features
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("24EU02012[AI&ML]-Correlation Heatmap")
plt.show()
```
