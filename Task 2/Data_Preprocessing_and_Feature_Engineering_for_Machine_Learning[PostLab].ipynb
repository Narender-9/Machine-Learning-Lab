```python
# @title Step 1: Import Libraries, Load Dataset, and Display Basic Information
import pandas as pd
import numpy as np
import seaborn as sns

# Load Titanic dataset
df = sns.load_dataset('titanic')

# Display first 5 rows
print(df.head())
```

```python
# Dataset information
print(df.info())

# Shape of dataset
print("Rows and Columns:", df.shape)
```

```python
# @title Step 2: Display Column Names, Check Missing Values, and Missing Percentage
# Column names
print(df.columns)

# Missing values
print(df.isnull().sum())

# Missing value percentage
missing = (df.isnull().sum() / len(df)) * 100
print(missing)
```

```python
# @title Step 3: Data Cleaning (Drop Column, Fill Missing Values, Remove Duplicates)
# Drop column with many missing values
df = df.drop('deck', axis=1)

# Fill missing values
df['age'] = df['age'].fillna(df['age'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df['embark_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])

# Check remaining missing values
print(df.isnull().sum())

# Remove duplicate rows
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
# @title Step 4: Check Data Types and Apply Label Encoding
# Check data types
print(df.dtypes)

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

categorical = df.select_dtypes(include=['object', 'category']).columns

for col in categorical:
    df[col] = le.fit_transform(df[col].astype(str))

# Display encoded dataset
print(df.head())
```

```python
# @title Step 5: Feature Scaling (Standardization)
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

num_cols = ['age', 'fare']

df[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
# @title Step 6: Display Preprocessed Dataset
print(df.head())
```

```python
# @title Step 7: Save the Preprocessed Dataset
df.to_csv("Titanic_Preprocessed.csv", index=False)
```
